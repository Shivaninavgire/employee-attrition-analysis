```python
total_employees = len(df)

employees_left = (df["Attrition"] == "Yes").sum()

attrition_rate = (employees_left / total_employees) * 100

print("Total Employees:", total_employees)
print("Employees Who Left:", employees_left)
print("Attrition Rate:", round(attrition_rate, 2), "%")
```

```python
employees_stayed = (df["Attrition"] == "No").sum()

retention_rate = (employees_stayed / total_employees) * 100

print("Employees Who Stayed:", employees_stayed)
print("Retention Rate:", round(retention_rate, 2), "%")
```

```python
average_income = df["MonthlyIncome"].mean()

print(
    "Average Monthly Income:",
    round(average_income, 2)
)
```

```python
average_years = df["YearsAtCompany"].mean()

print(
    "Average Years at Company:",
    round(average_years, 2)
)
```

```python
overtime_percentage = (
    (df["OverTime"] == "Yes").sum()
    / len(df)
) * 100

print(
    "Employees Working Overtime:",
    round(overtime_percentage, 2),
    "%"
)
```

```python
numeric_df = df.select_dtypes(include=np.number)
```

```python
df["AttritionBinary"] = df["Attrition"].map({
    "No": 0,
    "Yes": 1
})
```

```python
print("EMPLOYEE ATTRITION BUSINESS INSIGHTS")
print("=" * 50)

attrition_rate = (df['Attrition'] == 'Yes').mean() * 100

print(f"Overall Attrition Rate: {attrition_rate:.2f}%")

print("\nEmployees who left:")
print((df['Attrition'] == 'Yes').sum())

print("\nEmployees who stayed:")
print((df['Attrition'] == 'No').sum())

print("\nAverage Monthly Income:")
print(f"₹{df['MonthlyIncome'].mean():,.2f}")

print("\nAverage Years at Company:")
print(f"{df['YearsAtCompany'].mean():.2f} years")
```

```python
satisfaction_attrition = pd.crosstab(
    df['JobSatisfaction'],
    df['Attrition'],
    normalize='index'
) * 100

display(satisfaction_attrition)
```

```python
department_attrition = pd.crosstab(
    df['Department'],
    df['Attrition'],
    normalize='index'
) * 100

display(department_attrition)
```
